```python
# Import required libraries

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)
```

```python
# Dataset for house purchase prediction

data = {
    'Age': [22, 25, 28, 31, 35, 40, 45, 50,
            27, 33, 38, 29, 32, 26, 41, 36],

    'Income': [25, 30, 35, 45, 50, 60, 70, 80,
               40, 55, 65, 38, 30, 70, 35, 75],

    'Buy_House': [0, 0, 0, 1, 1, 1, 1, 1,
                  0, 1, 1, 0, 0, 1, 0, 1]
}

df = pd.DataFrame(data)

print("House Purchase Dataset:")
print(df)
```

```python
# Input variables
X = df[['Age', 'Income']]

# Target variable
y = df['Buy_House']

print("Input Variables:")
print(X)

print("\nTarget Variable:")
print(y)
```

```python
# Create Decision Tree Classifier

model = DecisionTreeClassifier(
    criterion='gini',
    max_depth=3,
    random_state=42
)

# Train the model
model.fit(X, y)

print("Decision Tree Model Created Successfully!")
```

```python
# Make predictions

predictions = model.predict(X)

print("Actual Values:")
print(list(y))

print("\nPredicted Values:")
print(list(predictions))
```

```python
# Calculate accuracy

accuracy = accuracy_score(y, predictions)

print("Model Accuracy:", accuracy)
print("Model Accuracy:", accuracy * 100, "%")
```

```python
# Display the Decision Tree

plt.figure(figsize=(18, 10), dpi=120)

plot_tree(
    model,
    feature_names=['Age', 'Income'],
    class_names=['No House', 'Buy House'],
    filled=True,
    rounded=True,
    fontsize=11
)

plt.title(
    "Decision Tree - House Purchase Prediction",
    fontsize=18
)

plt.tight_layout()
plt.show()
```

```python
# Calculate confusion matrix

cm = confusion_matrix(y, predictions)

print("Confusion Matrix:")
print(cm)
```

```python
# Display confusion matrix as a graph

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=['No House', 'Buy House']
)

fig, ax = plt.subplots(figsize=(7, 6))

disp.plot(
    ax=ax,
    values_format='d',
    cmap='Blues'
)

plt.title(
    "Confusion Matrix - House Purchase",
    fontsize=16
)

plt.show()
```

```python
# Test the model with new people

new_people = pd.DataFrame({
    'Age': [24, 34, 45, 29],
    'Income': [30, 55, 70, 65]
})

new_predictions = model.predict(new_people)

print("New People:")
print(new_people)

print("\nPredictions:")

for i in range(len(new_people)):
    if new_predictions[i] == 1:
        result = "Buy House"
    else:
        result = "No House"

    print(
        "Age =", new_people.iloc[i]['Age'],
        ", Income =", new_people.iloc[i]['Income'],
        "→", result
    )
```

```python
# Calculate accuracy

weather_accuracy = accuracy_score(
    y_weather,
    weather_predictions
)

print("Weather Model Accuracy:", weather_accuracy)
print(
    "Weather Model Accuracy:",
    weather_accuracy * 100,
    "%"
)
```
